```python
# External imports
from typing import *
from typing_extensions import *
from numbers import *
import unittest
import sys
import os
import random

# Add needed directories to the path
sys.path.append(os.path.join("..", "utils"))

# PyRat imports
from MinHeap import MinHeap
```

<h1 style="background-color: gray;
           color: black;
           padding: 20px;
           text-align: center;">DEFINE THE TESTS</h1>

The `unittest` library requires the creation of a class that extends `unittest.TestCase`. \
For each method to test, we need to define a method in the test class. \
Each of these test methods should call the tested method with various inputs to check that produced outputs match expected ones.

```python
class MinHeapTests (unittest.TestCase):

    """
        This class tests the methods of the MinHeap class.
        For each method, we test it with a few different configurations.
    """

    #############################################################################################################################################
    #                                                                 UNIT TESTS                                                                #
    #############################################################################################################################################

    def test_minheap(self: Self) -> None:
        """
            This method test the MinHeap class.
            It test:
            -   adding new items to empty heap
            -   poping an item
            -   replacing an already existing item
            -   not changing an item if its value is smaller than the new one
            -   test str and int as item key
        """
        # Create a min-heap
        heap = MinHeap()

        # Add elements to the heap
        heap.add_or_replace("A", 50)
        heap.add_or_replace("B", 22)
        heap.add_or_replace("C", 10)

        self.assertEqual(heap.elements, [("A", 50), ("B", 22), ("C", 10)])
        self.assertEqual(heap.remove(), ("C", 10))
        self.assertEqual(heap.elements, [("A", 50), ("B", 22)])

        # Try to add a greater value to an already existing element
        heap.add_or_replace("B", 45)
        self.assertEqual(heap.elements, [("A", 50), ("B", 22)])

        # Add a new element to the heap
        heap.add_or_replace("A", 35)
        self.assertEqual(sorted(heap.elements), sorted([("A", 35), ("B", 22)]))

        self.assertEqual(heap.remove(), ("B", 22))
        self.assertEqual(heap.elements, [("A", 35)])

        # Now with integer key
        heap = MinHeap()
        heap.add_or_replace(1, 40)
        heap.add_or_replace(5, 25)
        heap.add_or_replace(2, 30)
        heap.add_or_replace(4, 50)
        self.assertEqual(sorted(heap.elements), sorted([(1, 40), (5, 25), (2, 30), (4, 50)]))
        self.assertEqual(heap.remove(), (5, 25))
        self.assertEqual(sorted(heap.elements), sorted([(1, 40), (2, 30), (4, 50)]))
        self.assertEqual(heap.remove(), (2, 30))
        self.assertEqual(sorted(heap.elements), sorted([(1, 40), (4, 50)]))
        heap.add_or_replace(4, 20)
        self.assertEqual(sorted(heap.elements), sorted([(1, 40), (4, 20)]))
        self.assertEqual(heap.remove(), (4, 20))
```
